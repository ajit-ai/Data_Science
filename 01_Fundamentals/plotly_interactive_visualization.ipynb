Plotly charts are zoomable, hoverable, toggleable - ideal for EDA and dashboards. Plotly Express is the one-liner interface on top.

Uses built-in datasets (tips, gapminder, iris) so it runs anywhere.

## 1. Scatter with built-in interactivity

```python
import plotly.express as px

df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length",
                 color="species", size="petal_length",
                 hover_data=["species_id"], title="Iris - hover, box-zoom, click legend")
fig.show()
```

## 2. Bar / histogram / box

```python
tips = px.data.tips()
px.bar(tips.groupby(["day", "sex"], as_index=False).total_bill.sum(),
       x="day", y="total_bill", color="sex", barmode="group",
       title="Total bill by day & gender").show()
px.histogram(tips, x="tip", color="time", marginal="box",
             title="Tip distribution").show()
```

## 3. Facets + animation frames

```python
gap = px.data.gapminder()
px.scatter(gap, x="gdpPercap", y="lifeExp", size="pop",
           color="continent", hover_name="country", log_x=True,
           animation_frame="year", animation_group="country",
           size_max=55, range_y=[20, 90],
           title="Hans Rosling in 3 lines").show()
```

```python
px.line(gap[gap.country.isin(["India", "China", "United States"])],
        x="year", y="gdpPercap", color="country",
        line_shape="spline", markers=True, log_y=True,
        title="GDP per capita").show()
```

## 4. Choropleth map

```python
recent = gap[gap.year == 2007]
px.choropleth(recent, locations="iso_alpha", color="lifeExp",
              hover_name="country", color_continuous_scale="Viridis",
              title="Life expectancy 2007").show()
```

## Plotly vs matplotlib/seaborn
| | Matplotlib/Seaborn | Plotly |
|---|---|---|
| static reports/papers | best | export PNG |
| exploration, dashboards | ok | **best** |
| syntax | verbose-ish | `px.<chart>(df, ...)` |
| output | image | HTML widget (zoom/hover) |

Export standalone files with `fig.write_html("chart.html")`; embed in Streamlit via `st.plotly_chart`.
