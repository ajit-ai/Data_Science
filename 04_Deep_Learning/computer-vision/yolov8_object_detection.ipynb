## 1. Inference with pretrained weights

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2

model = YOLO("yolov8n.pt")            # nano; try s/m/l/x for accuracy
results = model.predict("https://ultralytics.com/images/bus.jpg", conf=0.4)

r = results[0]
img = cv2.cvtColor(r.plot(), cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 7)); plt.imshow(img); plt.axis("off")
plt.title("YOLOv8n predictions"); plt.show()

for b in r.boxes:
    print(f"{model.names[int(b.cls)]:<10} conf={float(b.conf):.2f}  xyxy={[round(v) for v in b.xyxy[0].tolist()]}")
```

## 3. Fine-tune on a custom dataset (structure + command)

```python
dataset_yaml = """
# data.yaml  - 3 folders with images + matching .txt labels
train: ../datasets/mydata/train/images
val:   ../datasets/mydata/valid/images
names:
  0: helmet
  1: vest
  2: person
"""
training_cmd = """
from ultralytics import YOLO
model = YOLO('yolov8n.pt')                    # transfer learning start point
model.train(data='data.yaml', epochs=50, imgsz=640, batch=16)
metrics = model.val()
"""
print(dataset_yaml); print(training_cmd)
```

## Cheat sheet
| Task | Call |
|---|---|
| detect | `predict()` |
| segment | load `yolov8n-seg.pt` |
| classify | `yolov8n-cls.pt` |
| pose | `yolov8n-pose.pt` |
| export | `model.export(format='onnx')` |

Labeling tools: Roboflow / Label Studio / CVAT export straight to YOLO txt format.
