Ollama runs open models (Llama, Qwen, Mistral, Gemma) entirely on your machine - or inside a free Colab VM. No API keys, no rate limits, data never leaves the box.

This notebook installs Ollama **inside Google Colab**, pulls a small model and chats with it over HTTP.

## 1. Install Ollama in the Colab VM (~1 min)

```python
!curl -fsSL https://ollama.com/install.sh | sh
```

## 2. Start the server in the background

```python
import subprocess, time, os

server = subprocess.Popen(["ollama", "serve"],
                          env={**os.environ, "OLLAMA_HOST": "127.0.0.1:11434"})
time.sleep(5)
print("server pid:", server.pid)
```

## 3. Pull a small model (llama3.2 1B ~= 1.3 GB)

```python
!ollama pull llama3.2:1b
!ollama list
```

## 4. Chat over the REST API

```python
import requests, json

URL = "http://127.0.0.1:11434/api/chat"

def chat(messages, stream=False):
    r = requests.post(URL, json={"model": "llama3.2:1b",
                                 "messages": messages, "stream": stream})
    return r.json()["message"]["content"]

messages = [{"role": "system", "content": "Answer in one short sentence."},
            {"role": "user", "content": "Why does attention beat RNNs for long text?"}]
print(chat(messages))
```

## 5. Structured output (JSON mode)

```python
r = requests.post(URL, json={
    "model": "llama3.2:1b",
    "messages": [{"role": "user",
                  "content": "List 2 planets as JSON array of {name, moons}"}],
    "format": "json", "stream": False})
print(json.dumps(json.loads(r.json()["message"]["content"]), indent=2))
```
