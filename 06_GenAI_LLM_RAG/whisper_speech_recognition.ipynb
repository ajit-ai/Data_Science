## 1. Get a sample audio clip (public domain, LibriVox)

```python
import urllib.request

url = "https://www.mfiles.co.uk/mp3-downloads/bach-minuet-in-g-major.mp3"
urllib.request.urlretrieve(url, "sample.mp3")

# Any mp3/wav/m4a works. To use your own voice: record and upload to Colab.
import IPython
IPython.display.Audio("sample.mp3")
```

## 2. Transcribe

```python
import whisper

model = whisper.load_model("base")     # tiny<base<small<medium<large-v3
result = model.transcribe("sample.mp3")
print("detected language:", result["language"])
print("text:", result["text"][:300])
```

## 3. Timestamped segments

```python
for seg in result["segments"][:5]:
    print(f"[{seg['start']:6.2f} -> {seg['end']:6.2f}] {seg['text']}")
```

## 4. Model size vs accuracy tradeoff

```python
tiny_model = whisper.load_model("tiny")
r_tiny = tiny_model.transcribe("sample.mp3")
print("tiny :", r_tiny["text"][:120])
print("base :", result["text"][:120])
```

## Practical notes
- `language="en"` skips auto-detect -> faster.
- Non-English audio: Whisper handles 90+ languages; use `large-v3` for best multilingual accuracy.
- Word timestamps: `word_timestamps=True`.
- Long meetings: chunk by silence (e.g., pydub) then merge segments.
- Pipeline idea: mic audio -> Whisper -> LLM summary (combine with `ollama_local_llm.ipynb`).
